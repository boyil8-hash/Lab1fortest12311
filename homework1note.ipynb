```python
import numpy as np  # import numpy for computation 
import matplotlib  # import matplot for better graphing  
import matplotlib.pyplot as plt  # sub liberary under matplot 
import scipy # this liberary is for stats
from scipy import stats
# set the size of the plot to something useful 
plt.rcParams["figure.figsize"] = (15 , 10)  # size of the plot 
```

```python
d = stats.norm.rvs(loc = 5.0, scale = .01, size=1000000) # random choose 1000000 number from a normal distribution with median at 5.0, standard deviation at 0.01
```

```python
fig, ax = plt.subplots(1,1) # create 1 row 1 column graph 
ax.hist( d,50, density=True ) # create a density graph  of d with 100 bins
plt.tick_params(labelsize =24) # font size of text on x/y axis is 24 
plt.xlim([4.95,5.05]) # x limit from 4.95 to 5.05 which is u+- 5sd 
x = np.linspace(4.95, 5.05,1000) # take 10000 points from 4.95 to 5.05 on x axis, so that the graph looks smooth
ax.plot(x, stats.norm.pdf(x,loc = 5., scale = 0.01), linewidth =8, alpha =0.7)  # compute every probability density over xand graph, linewideth is 8, alpha is opacity
plt.show() # show the plot 
```

```python
ax = plt.hist(d,50, density = True) 
plt.yscale("log") 
plt.tick_params(labelsize =24) 
plt.xlim([4.95,5.05]) 
plt.show()
```
